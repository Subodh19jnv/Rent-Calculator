```python
rent = float(input("Enter House Rent: "))
electricity = float(input("Enter Electricity Bill: "))
water = float(input("Enter Water Bill: "))
internet = float(input("Enter Internet Bill: "))
other = float(input("Enter Other Expenses: "))

members = int(input("Enter Number of Flatmates: "))

total = rent + electricity + water + internet + other

share = total / members

print("\n----- RENT SUMMARY -----")

print("House Rent:", rent)
print("Electricity Bill:", electricity)
print("Water Bill:", water)
print("Internet Bill:", internet)
print("Other Expenses:", other)

print("Total Expenses:", total)
print("Each Person Pays:", round(share, 2))
```
